```python
import os
import shutil

# -----------------------------
# 1. Define directories
# -----------------------------
source_directory = "/content/messy_movie.file2"
new_directory = "/content/renamed_movie.file"

# -----------------------------
# 2. Create folders if not exist
# -----------------------------
if not os.path.exists(source_directory):
    os.makedirs(source_directory)
    print("Source directory created")
else:
    print("Source directory found")

if not os.path.exists(new_directory):
    os.makedirs(new_directory)
    print("New directory created")
else:
    print("New directory found")

# -----------------------------
# 3. Create messy movie files
# (only for Colab testing)
# -----------------------------
messy_files = [

 "Inception--2010--Nolan.mp4",
  "THE-GODFATHER_1972_Ford.avi",
  "star_wars!1977&Lucas.avi",
  "Harry-Potter-2001.Rowling.mp4",
  "AVATAR!2009_Cameron.mp4",
  "Dark_Knight;2008!Nolan.mp4",
  "Titanic_1997-Cameron.avi",
  "matrix-1999&wachowski.mp4",
  "Shawshank-Redemption;1994,Robbins.avi",
  "Forrest_Gump-1994.TomHanks.avi",
  "Gladiator;2000_Ridley!Scott.mp4",
  "Lord_of_Rings-2003-Jackson.mp4",
  "Pulp-Fiction_1994_Tarantino.avi",
  "Interstellar--2014--Nolan.mp4",
  "Back;ToTheFuture;1985;Spielberg.avi",
  "The_Lion!King!1994;Disney.flv",
  "Frozen-2013_Disney.mp4",
  "Pirates!of_the-Caribbean-2003-Verbinski.mp4",
  "Jurassic;Park-1993-Spielberg.flv",
  "The_Avengers;2012-Whedon.mp4"
]

for file in messy_files:
    open(os.path.join(source_directory, file), "w").close()

print("Messy files added\n")

# -----------------------------
# 4. Function to rename and move
# -----------------------------
def RenameMoveFiles(filename):
    print(f"Old movie name - {filename}")

    # split filename and extension
    name, ext = os.path.splitext(filename)

    # clean unwanted characters
    name = name.replace("_", " ")
    name = name.replace("!", " ")
    name = name.replace("&", " ")
    name = name.replace("-", " ")
    name = name.replace(".", " ")
    name = name.replace(";", " ")
    name = name.replace("--", " ")

    # remove extra spaces
    parts = name.split()
    clean_name = " ".join(parts)

    new_filename = clean_name + ext

    print(f"New movie name - {new_filename}\n")

    # full paths
    old_path = os.path.join(source_directory, filename)
    new_path = os.path.join(new_directory, new_filename)

    # move + rename
    shutil.move(old_path, new_path)

# -----------------------------
# 5. Process all files
# -----------------------------
for file in os.listdir(source_directory):
    RenameMoveFiles(file)

# -----------------------------
# 6. Final output
# -----------------------------
print("Final renamed files:")
print(os.listdir(new_directory))
```
